# Clinical NLU Playground

Interactive playground for testing `SearchService.parse_utterance(...)` against the bundled `gosmart-clinical-nlu` intent models.

## 1. Environment & SearchService Setup

```python
import sys
from pathlib import Path

# Resolve project root whether Jupyter is launched from the repo root or notebooks/
_cwd = Path.cwd().resolve()
PROJECT_ROOT = _cwd if (_cwd / 'src').is_dir() else _cwd.parent
SRC_DIR = PROJECT_ROOT / 'src'
for _p in (SRC_DIR, PROJECT_ROOT):
    if str(_p) not in sys.path:
        sys.path.insert(0, str(_p))

from gosmart_graphite.service.search_service import SearchService

search_svc = SearchService(require_nlu_model=True)

print(f'Project Root : {PROJECT_ROOT}')
print(f'NLU Ready    : {search_svc.nlu_ready}')
print(f'Model Bundle : {search_svc.nlu_model_path}')
print(f'Intents      : {list(search_svc.intent_models.names) if search_svc.intent_models else []}')
```

## 2. Test NLU Intents

```python
m = search_svc.parse_utterance('Show radiology report of accession 323333')
print(m)
```
